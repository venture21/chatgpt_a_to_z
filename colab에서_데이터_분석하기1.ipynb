```python
# csv파일을 불러와서 제일 앞에 5개 행을 보여줘

import pandas as pd
df = pd.read_csv('/content/korean-idol.csv')
df.head()
```

```python
# csv에서 읽어온 사람들 중에 키가 제일 큰 사람과 작은 사람을 알려줘

tallest_person = df.loc[df['키'].idxmax()]
shortest_person = df.loc[df['키'].idxmin()]

print(f"키가 가장 큰 사람: {tallest_person['이름']} (키: {tallest_person['키']}cm)")
print(f"키가 가장 작은 사람: {shortest_person['이름']} (키: {shortest_person['키']}cm)")
```

```python
# 혈액형이 A형인 사람의 목록을 출력해줘

blood_type_a_people = df[df['혈액형'] == 'A']
print("혈액형이 A형인 사람:")
for name in blood_type_a_people['이름']:
    print(name)
```

```python
# 키컬럼에 NaN이 있는데 평균값으로 채워줘

mean_height = df['키'].mean()
df['키'] = df['키'].fillna(mean_height)

print(f"키 컬럼의 평균값: {mean_height:.2f}")
print("NaN 값 처리 후 '키' 컬럼의 정보:")
df.info()
```

```python
display(df)
```

```python
df_sorted_by_height = df.sort_values(by='키')
display(df_sorted_by_height)
```

```python
# 그럼 수정된 내용을 result.csv파일로 저장해줘

df.to_csv('result.csv', index=False)
print("DataFrame이 'result.csv' 파일로 저장되었습니다.")
```
